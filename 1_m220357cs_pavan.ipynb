**i)Import Dermatology Dataset** 

```python
# Importing the necessary libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

missing_values =["N/a","na",np.nan]
dermatology_df = pd.read_csv("./dermatology.csv",na_values=missing_values)
dermatology_df
```

**ii)Data Cleaning**


```python
# Use an appropriate method to handle missing value
dermatology_df.dropna(how="all")
mean_values = dermatology_df.mean()
dermatology_df.fillna(mean_values,inplace=True)
print(dermatology_df.isnull().sum())
```

**iii)Feature engineering**

**iv)Exploratory data analysis (EDA)**

```python
import matplotlib.pyplot as plt
for column in dermatology_df.columns:
    plt.hist(dermatology_df[column])
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.title("Histogram of "+ column)
    plt.show()
```

**#Derive variance**

```python
mean_columns=dermatology_df.mean()
deviation = dermatology_df - mean_columns
deviation2= deviation**2
sum_deviation2 = deviation2.sum()
variances = sum_deviation2/(dermatology_df.count()-1)
print(variances)
```

**#Correlation coefficient matrix**

```python
correlation_matrix = dermatology_df.corr()
print(correlation_matrix)
```

**#Density plots of all features**

```python
import seaborn as sns
import matplotlib.pyplot as plt

for column in dermatology_df.columns[:-1]:
    figure,axis = plt.subplots()
    sns.kdeplot(dermatology_df[column],ax=axis)
    axis.set_title(f"Density plot of {column}")
    axis.set_xlabel(column)
    axis.set_ylabel("Density")
    plt.show()
```

**#Box plots of all features**

```python
import pandas as pd
import matplotlib.pyplot as plt

for column in dermatology_df.columns[:-1]:
    figure, axis = plt.subplots()
    axis.boxplot(dermatology_df[column])
    axis.set_title(f"Box plot of {column}")
    axis.set_xlabel(column)
    axis.set_ylabel("Value")
    plt.show()
```
