## **A simple object detection model using Faster R-CNN**

# **Importing necessary libraries**

```python
import torch
import torchvision
import numpy as np
import cv2
from matplotlib import pyplot as plt
import matplotlib
matplotlib.use('TkAgg')
```

# **Loading the pre-trained Faster R-CNN model:**

```python
model = torchvision.models.detection.fasterrcnn_resnet50_fpn(pretrained=True)
model.eval()
```

# **Loading an example image and preprocessing it:**

```python
img_path = './humans1.jpg'
img = cv2.imread(img_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img = img.astype(np.float32) / 255.0
img_tensor = torch.tensor(img, dtype=torch.float32).permute(2, 0, 1)
```

# **Feeding the image through the model and obtaining the object predictions:**

```python
outputs = model([img_tensor])
pred_boxes = outputs[0]['boxes'].detach().numpy()
pred_labels = outputs[0]['labels'].detach().numpy()
pred_scores = outputs[0]['scores'].detach().numpy()
```

# **Visualizing the object predictions by drawing bounding boxes around the detected objects:**

```python

fig, ax = plt.subplots()
ax.imshow(img)
for box, label, score in zip(pred_boxes, pred_labels, pred_scores):
    if score > 0.5:
        x1, y1, x2, y2 = box.astype(np.int32)
        label_text = str(label.item())
        bbox = {"facecolor": "white", "alpha": 0.7, "pad": 0.2}
        ax.add_patch(plt.Rectangle((x1, y1), x2 - x1, y2 - y1, linewidth=1, edgecolor='red', facecolor='none'))
plt.show()
```
